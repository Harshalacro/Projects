```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="whitegrid")
%matplotlib inline
```

```python
# Load the CSV file
df = pd.read_csv('creditcard.csv')  # make sure the file is in the same directory or update path

# Display first few rows
df.head()
```

```python
# Shape of the dataset
print(f"Shape: {df.shape}")

# Info and data types
df.info()

# Check for missing values
df.isnull().sum().any()
```

```python
# Count of each class
class_counts = df['Class'].value_counts()
print(class_counts)

# Plot the class distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Class')
plt.title('Distribution of Fraud (1) vs. Non-Fraud (0)')
plt.show()

# Percentage of fraud
fraud_percentage = class_counts[1] / class_counts.sum() * 100
print(f"Fraudulent transactions: {fraud_percentage:.4f}%")
```

```python
# Describe entire dataset
df.describe()

# Compare fraud vs. non-fraud stats
fraud_df = df[df['Class'] == 1]
non_fraud_df = df[df['Class'] == 0]

print("Fraudulent Transactions Summary:")
print(fraud_df[['Amount', 'Time']].describe())

print("\nNon-Fraudulent Transactions Summary:")
print(non_fraud_df[['Amount', 'Time']].describe())
```

```python
# Amount
plt.figure(figsize=(10, 4))
sns.histplot(non_fraud_df['Amount'], bins=100, color='blue', label='Non-Fraud', alpha=0.5)
sns.histplot(fraud_df['Amount'], bins=100, color='red', label='Fraud', alpha=0.5)
plt.legend()
plt.title('Transaction Amount Distribution')
plt.xlim([0, 2000])
plt.show()

# Time
plt.figure(figsize=(10, 4))
sns.kdeplot(non_fraud_df['Time'], label='Non-Fraud')
sns.kdeplot(fraud_df['Time'], label='Fraud')
plt.title('Transaction Time Distribution')
plt.legend()
plt.show()
```

```python
# Compute correlation matrix
corr_matrix = df.corr()

# Plot heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap='coolwarm', linewidths=0.5)
plt.title('Feature Correlation Matrix')
plt.show()

# Correlation with target
corr_target = corr_matrix['Class'].sort_values(ascending=False)
print("Top correlations with Class:\n", corr_target.head(10))
```

```python
# Boxplot of Amount by Class
plt.figure(figsize=(6, 4))
sns.boxplot(x='Class', y='Amount', data=df)
plt.title('Transaction Amount by Class')
plt.yscale('log')
plt.show()
```
